### Try and except

try блок, в котором мы «пытаемся выполнить» код.

except блок, который выполнится, если в try возникло исключение.

В Python исключение это особый сигнал, что что-то пошло не так (например, деление на 0 или наша искусственная ошибка)

```python
try:
    for i in range(5):
        if i == 3:
            raise StopIteration
        print(i)
except StopIteration:
    print("Цикл прерван через исключение")
```

### yield vs return

yield  это как return, но не завершает функцию

Когда функция встречает yield, она возвращает значение текущей итерации, но сохраняет своё состояние

```python
def gen_numbers(n):
    for i in range(n):
        return i

for x in gen_numbers(5):
    print(x)
```

```python
def gen_numbers(n):
    for i in range(n):
        yield i

for x in gen_numbers(5):
    print(x)
```

### Вложенные функции

```python
def factorial(number):
    if not isinstance(number, int):
        raise TypeError("Число должно быть целым.")
    if number < 0:
        raise ValueError("Число должно быть неотрицательным.")
    def inner_factorial(number):
        if number <= 1:
            return 1
        return number * inner_factorial(number - 1)
    return inner_factorial(number)
```

## декораторы функций

```python
def my_decorator(func):
    def wrapper():
        print("Что-то происходит ДО вызова функции.")
        func()
        print("Что-то происходит ПОСЛЕ вызова функции.")
    return wrapper

@my_decorator
def say_hello():
    print("Привет!")

say_hello()
```

```python
from functools import wraps

def uppercase(func):
    @wraps(func)
    def wrapper():
        return func().upper()
    return wrapper

@uppercase
def greet():
    return "Hello!"

print(greet())         
print(greet.__wrapped__()) 
```

```python
greet()
```

```python
greet.__wrapped__()
```

## Фибоначчи

```python
def fib(n):
    F = [-1]*(n+1)
    F[0] = 0
    F[1] = 1
    for i in range(2, n+1):
        F[i] = F[i - 1] + F[i - 2]
    return F[n]
```

## Кузнечик

Решите задачу о количестве способов достичь точки n из точки 1, если кузнечик умеет прыгать +1, +2 и +3.

```python
def count_ways(n):
    if n < 1:
        return 0
    if n == 1:
        return 1

    dp = [0] * (n + 1)
    dp[1] = 1
    if n >= 2:
        dp[2] = 1
    if n >= 3:
        dp[3] = 2

    for i in range(4, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2] + dp[i - 3]

    return dp[n]
```

```python
Решите задачу о количестве способов достичь точки n из точки 1, если кузнечик умеет прыгать +1, +2 и *3
```

```python
def count_ways(n):
    if n < 1:
        return 0

    dp = [0] * (n + 1)
    dp[1] = 1

    for i in range(2, n + 1):
        ways = 0
        if i - 1 >= 1:
            ways += dp[i - 1]
        if i - 2 >= 1:
            ways += dp[i - 2]
        if i % 3 == 0 and i // 3 >= 1:
            ways += dp[i // 3]
        dp[i] = ways

    return dp[n]
```

### еще кузнечик

В качестве примера задачи на нахождение наилучшего способа рассмотрим следующую задачу о кузнечике.

Пусть кузнечик прыгает на одну или две точки вперед, а за прыжок в каждую точку необходимо заплатить определенную стоимость, различную для различных точек. Стоимость прыжка в точку i задается значением P[i] списка стоимостей P. Необходимо найти минимальную стоимость маршрута кузнечика из точки 1 в точку n (то есть, наилучшее решение в данном случае — это маршрут с минимальной стоимостью).

Пусть F(n) — минимальная стоимость пути из 1 в n. Выведем рекуррентное соотношение для F(n). Чтобы попасть в точку n мы должны попасть в точку n-1 или n-2. Минимальные стоимости этих маршрутов будут равны F(n-1) и F(n-2) соответственно. К ним необходимо добавить значение P[n] за прыжок в клетку n. При этом из двух клеток n-1 и n-2 нужно выбрать тот маршрут, который имеет наименьшую стоимость.

Получим соотношение для F(n):

```python
5
1 3 2 4 5
```

```python
n = int(input())
P = [0] + list(map(int, input().split()))

F = [0] * (n + 1)
prev = [0] * (n + 1) 
F[1] = P[1]
prev[1] = 0

if n >= 2:
    F[2] = P[1] + P[2]
    prev[2] = 1

for i in range(3, n + 1):
    if F[i - 1] <= F[i - 2]:
        F[i] = F[i - 1] + P[i]
        prev[i] = i - 1
    else:
        F[i] = F[i - 2] + P[i]
        prev[i] = i - 2

path = []
i = n
while i > 0:
    path.append(i)
    i = prev[i]

path.reverse()

print("Минимальная стоимость:", F[n])
print("Путь:", " ".join(map(str, path)))
```

## рюкзак

```python
W = 10
w = [6, 5, 5]
v = [6, 5, 5]
```

```python
def greedy(W, w, v):
    items = sorted(zip(w, v), key=lambda x: x[1] / x[0], reverse=True)
    total_v = 0
    for weight, value in items:
        if weight <= W:
            W -= weight
            total_v += value
    return total_v


def knapsack(W, w, v):
    n = len(w)
    dp = [0] * (W + 1)
    for i in range(n):
        for cap in range(W, w[i] - 1, -1):
            dp[cap] = max(dp[cap], dp[cap - w[i]] + v[i])
    return dp[W]
```

```python
print("Жадный:  ", greedy(W, w, v))
print("динамика: ", knapsack(W, w, v))
```
